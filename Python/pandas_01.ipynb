*** Pandas 실무에서 사용목적
1. 테이블형 데이터를 다루는 데 있어서 압도적인 기능
2. 원본 데이터(raw data)의 전처리(preprocessing) 및 정제(cleansing) 작업 시, 비교적 간단한 코드만으로 가능
3. 데이터 분석 기법을 활용할 때에 굉장히 많이 등장
    - EDA : 탐색적 데이터 분석 = Exploratory Data Analysis
      > 본격적인 데이터 분석 전, 데이터 사이즈, 결측치, 이상치, 평균·중앙값, 사분위 데이터 등을 통해서 데이터의 패턴을 파악하기 위한 목적
    - LDA : 잠재적 주제 도출 분석 과정 = Latent Dirichlet Allocation = 잠재 디리클레 할당
      > EDA 과정을 걸쳐서 충분히 분석할만한 가치가 있다고 판단은 하였으나, 파편화되어있는 데이터들이 어떤 주제로 구성되어 있는지를 분류·파악하기 위한 목적
    - ETL : 데이터 추출 처리 분석 = Extract Transform Load
      > 원본 데이터 = 원시데이터는 정보가(데이터의 패턴 조회, 가설 설정) 아닙니다. -> 원시 데이터를 유용한 정보성 요소로 변환시키도록 하기 위한 분석 과정
      > 데이터의 패턴을 조회하고 여기에 대한 가설을 설정하면 정보가 됨
